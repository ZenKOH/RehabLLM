# Tokenisation

Inspect how rehabilitation terminology is segmented by the trained SentencePiece model.


```python
from rehab_minillm.tokenizer import SentencePieceTokenizer

tok = SentencePieceTokenizer("../data/processed/rehab_sp.model")
text = "Robot-assisted upper-limb rehabilitation after stroke"
ids = tok.encode(text, add_bos=True, add_eos=True)
print(ids)
print(tok.decode(ids))
```
